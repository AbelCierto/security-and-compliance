> El LLM no debe ser el componente que decide qué información puede ver un usuario.

## Sección 1 – Contexto arquitectónico de BankAssist HR

BankAssist HR es un asistente interno para empleados de un banco latinoamericano.

El sistema permite consultar información relacionada con:

- Políticas de vacaciones y licencias.
- Beneficios y prestaciones sociales.
- Procedimientos internos de Recursos Humanos.
- Manuales operativos básicos.
- Información organizacional autorizada.

La solución utiliza una arquitectura basada en recuperación aumentada por generación, o **RAG**:

1. El empleado realiza una pregunta desde la intranet.
2. El backend autentica al usuario.
3. El sistema determina su rol y atributos de acceso.
4. El módulo RAG busca información únicamente en dominios autorizados.
5. El sistema filtra los documentos según su clasificación.
6. El contexto permitido se envía al LLM.
7. El modelo genera una respuesta.
8. La respuesta se valida antes de mostrarse.
9. La operación se registra mediante logging mínimo y seguro.

### Restricción principal

El sistema no debe enviar todos los documentos disponibles al LLM para intentar filtrar la información después.

La autorización debe aplicarse antes de:

- Recuperar documentos.
- Construir el contexto.
- Construir el prompt.
- Invocar el modelo.

> El modelo debe operar sobre un subconjunto de datos previamente autorizado por la arquitectura.

## Sección 2 – Flujo seguro de una solicitud

Completa o adapta el siguiente flujo para representar cómo debería procesarse una solicitud de forma segura:

```text
Empleado
   |
   v
Portal interno
   |
   v
Autenticación fuerte y emisión de identidad
   |
   v
Validación de rol, atributos, propósito y riesgo de sesión
   |
   v
Aplicación de políticas de acceso (RBAC + ABAC)
   |
   v
Selección del dominio e índice autorizado
   |
   v
Filtro obligatorio por clasificación, vigencia y permisos
   |
   v
Recuperación de contexto permitido
   |
   v
Minimización, detección de instrucciones maliciosas y protección del contexto
   |
   v
Invocación cifrada del LLM sin entrenamiento del proveedor
   |
   v
Validación de seguridad, citas y alcance de la respuesta
   |
   v
Respuesta al empleado
   |
   v
Logging mínimo, auditoría, monitoreo y aplicación de retención
```

### Tarea arquitectónica

| Control | Punto del flujo donde se aplica | Justificación |
|---|---|---|
| Autenticación | Antes de crear la solicitud en el backend | Confirma la identidad mediante SSO/MFA y evita consultas anónimas. |
| RBAC | Backend y motor de políticas, antes de seleccionar índices | Concede permisos base por función sin delegar autorización al LLM. |
| ABAC | Motor de políticas, antes de recuperar documentos | Evalúa red, dispositivo, horario, propósito, sensibilidad y riesgo de sesión. |
| Filtrado de contexto | Consulta al índice, antes de devolver chunks | Impide que un documento no autorizado llegue al prompt o a una caché. |
| Minimización de datos | Normalización de la consulta y construcción del contexto | Envía solo los campos y fragmentos necesarios para responder. |
| Cifrado | TLS/mTLS en tránsito y KMS en reposo | Reduce exposición de identidad, contexto y secretos entre componentes. |
| Validación de respuesta | Después de la generación y antes de mostrarla | Detecta PII, contenido fuera de alcance, prompt injection y respuestas sin soporte. |

## Investigación guiada 1 – Diseño de arquitectura segura

1. **El acceso debe preceder al retrieval:** la similitud semántica mide relevancia, no autorización. Si el filtro se aplica después, el chunk ya pudo quedar expuesto al proceso, al prompt, a la caché o a los logs.
2. **Autenticación versus autorización:** autenticar prueba quién es el sujeto; autorizar decide qué acción puede realizar sobre un recurso, bajo qué propósito y contexto.
3. **Riesgos de un índice global:** mezcla dominios con distintas clasificaciones, amplía el impacto de una credencial comprometida, dificulta el aislamiento, favorece errores de filtros y hace más difícil demostrar qué corpus podía consultar cada rol.
4. **Evidencia del filtro:** conservar `request_id`, usuario anonimizado, rol/atributos relevantes, dominio e índice seleccionado, versión de política, decisión, IDs/versiones de documentos devueltos, resultado y hash de integridad. No conservar el contenido completo salvo obligación aprobada.

## Sección 4 – Clasificación de datos y dominios

Antes de diseñar controles de acceso, el arquitecto debe identificar qué tipos de datos procesa la aplicación, qué tan sensibles son y a qué dominio pertenecen.

En BankAssist HR no todos los documentos tienen el mismo nivel de riesgo ni deben estar disponibles para los mismos usuarios.

### Conceptos de diseño

- **Dominio:** área funcional o de negocio a la que pertenece la información.
- **Clasificación:** nivel de sensibilidad o impacto asociado al dato.
- **Propietario del dato:** área responsable de definir quién puede acceder.
- **Propósito de uso:** finalidad autorizada para procesar la información.
- **Ciclo de vida:** tiempo durante el cual el dato puede conservarse.
- **Regla de acceso:** condición que debe cumplirse para consultar el dato.

### Posibles dominios de BankAssist HR

- RRHH general.
- Compensación y beneficios.
- Salud ocupacional.
- Relaciones laborales.
- Información organizacional.
- Auditoría y cumplimiento.
- Operaciones internas.

> La clasificación no debe realizarse solo por el formato del archivo. Un PDF, un embedding o un log también pueden contener información sensible y deben estar sujetos a controles.

## Metadatos obligatorios para documentos RAG

Los metadatos obligatorios son:

```text
document_id, chunk_id, title, domain, data_owner, classification,
allowed_roles, allowed_attributes, purpose, source_system,
created_at, updated_at, expires_at, version, embedding_version,
retention_class, revocation_status, integrity_hash
```

`domain` y `classification` seleccionan el índice y el filtro de seguridad; `allowed_roles` y `allowed_attributes` permiten RBAC/ABAC; `purpose` limita el uso; `expires_at` y `revocation_status` excluyen documentos no vigentes; `source_system`, `version` e `integrity_hash` reconstruyen el origen y evitan usar una copia alterada; `retention_class` gobierna el ciclo de vida. Solo el propietario del dato puede proponer cambios y seguridad/privacidad los aprueban. Cada modificación queda versionada y auditable.

## Decisión arquitectónica – Índices vectoriales por dominio

La propuesta de índice global con filtrado posterior es insegura: el control de acceso ocurre demasiado tarde, porque los chunks ya pudieron exponerse al RAG, al prompt, a caches, a trazas o al propio modelo. El LLM podría recibir salarios, salud ocupacional o auditoría aunque no los muestre. También se pierde aislamiento, aumenta el radio de impacto y se dificulta probar qué corpus estaba autorizado.

La alternativa es separar índices por dominio y aplicar filtros obligatorios previos por identidad, rol, atributos, clasificación, propósito, vigencia y revocación. La auditoría conserva `request_id`, política, índice seleccionado, IDs/versiones de documentos, decisión y resultado, sin guardar el contenido completo.

**Alternativa seleccionada:**

- [ ] Índice global con filtrado posterior.
- [ ] Índice global con filtros obligatorios previos.
- [ ] Índices separados por dominio.
- [x] Índices separados por dominio y filtros obligatorios previos.

**Justificación arquitectónica:** Esta opción aplica defensa en profundidad, limita el radio de impacto de una credencial comprometida, permite mínimo privilegio y produce evidencia clara de que el contexto fue autorizado antes de la inferencia.

## Preguntas reflexivas

### Reflexión 1

Un embedding debe considerarse sensible porque puede permitir recuperación, inferencia, correlación o reconstrucción parcial del documento original. Por eso hereda como mínimo la clasificación, permisos, retención y requisitos de borrado de su fuente.

### Reflexión 2

Si no se clasifican embeddings, caches y logs, una persona podría recuperar información por una ruta secundaria aunque el documento original esté protegido. También se perdería el borrado completo y la auditoría. La clasificación debe propagarse a todas las copias derivadas y sus accesos deben ser independientes.

## Preguntas que podría hacer un auditor

La organización mantiene un catálogo con propietario, dominio, clasificación, vigencia y hash. El backend aplica filtros antes del retrieval; los embeddings heredan controles de la fuente y se eliminan con ella. Solo propietarios aprobados modifican metadatos, y cada cambio tiene revisión y versión. Los documentos sin metadatos se rechazan. Los eventos conservan IDs/versiones de documentos y política, aplican TTL a documentos, embeddings, caches y logs, y permiten demostrar qué fue utilizado sin guardar contenido sensible.

## Sección 5 – Diseño de control de acceso

En una aplicación AI/LLM, el control de acceso debe determinar qué información puede llegar al sistema de inferencia.

No es suficiente con controlar únicamente el acceso al portal. También deben protegerse:

- Los documentos originales.
- Los fragmentos utilizados por el RAG.
- Los índices vectoriales.
- Los embeddings.
- El contexto recuperado.
- Las conversaciones.
- Los logs y registros de auditoría.
- Las herramientas internas, si el sistema incorpora agentes.

### Principio de mínimo privilegio

Cada usuario, servicio o componente debe tener únicamente los permisos necesarios para cumplir su función.

En BankAssist HR, el sistema debe:

1. Autenticar al usuario.
2. Obtener su identidad, rol y atributos.
3. Evaluar la política de acceso.
4. Consultar únicamente dominios autorizados.
5. Recuperar solo documentos permitidos.
6. Enviar al LLM únicamente el contexto autorizado.

> El LLM no debe utilizarse como mecanismo de autorización.

## RBAC – Control basado en roles

RBAC (*Role-Based Access Control*) asigna permisos según el rol organizacional del usuario.

### Roles iniciales de BankAssist HR

- **Empleado:** consulta políticas generales autorizadas.
- **Supervisor:** consulta políticas generales y ciertos procedimientos de supervisión.
- **RRHH:** consulta y administra información de Recursos Humanos según sus funciones.
- **Auditor:** consulta evidencias y registros autorizados, pero no necesariamente el contenido completo de todos los documentos.
- **Administrador técnico:** administra componentes técnicos, pero no debería obtener automáticamente acceso al contenido de negocio.
- **Servicio RAG:** recupera documentos según las políticas evaluadas; no decide por sí mismo qué usuario tiene acceso.

### Advertencia arquitectónica

Un rol técnico no debe equivaler automáticamente a un permiso para consultar información sensible.

Por ejemplo, un administrador de infraestructura podría administrar una base de datos, pero no debería tener acceso ilimitado al contenido de licencias médicas o información salarial.

### Tarea

Identifica qué permisos necesita cada rol y qué permisos deben estar explícitamente prohibidos.

## ABAC – Control basado en atributos y contexto

ABAC (*Attribute-Based Access Control*) complementa RBAC utilizando atributos del usuario, del recurso y del contexto.

### Posibles atributos del usuario

- Rol.
- Área o departamento.
- Ubicación.
- Nivel de autorización.
- Tipo de contrato.
- Estado del usuario.
- Participación en un caso o proceso.

### Posibles atributos del recurso

- Dominio.
- Sensibilidad.
- Propietario.
- Tipo de información.
- Propósito de uso.
- Fecha de expiración.

### Posibles atributos del contexto

- Hora de acceso.
- Red corporativa o externa.
- Dispositivo administrado o no administrado.
- Nivel de riesgo de la sesión.
- Estado de una aprobación.
- Entorno de ejecución.

### Ejemplos de políticas

- Un documento financiero puede consultarse únicamente durante horario laboral.
- Un registro de licencia médica solo puede ser consultado por RRHH autorizado.
- Una consulta sensible puede requerir red corporativa y dispositivo administrado.
- Un usuario puede acceder a un dominio, pero no exportar su contenido.

## Matriz ABAC de decisiones contextuales

| No. | Situación | Atributos que deben evaluarse | Decisión esperada | Justificación |
|---|---|---|---|---|
| 1 | Empleado consulta una política general desde la intranet | Cuenta activa, rol empleado, red corporativa, propósito de consulta, clasificación interna | Permitir | Es información interna y el contexto coincide con la política. |
| 2 | Supervisor consulta información salarial desde una red externa | Rol, dominio, autorización salarial, red, dispositivo, horario, caso aprobado | Denegar o exigir step-up y aprobación | El rol por sí solo no autoriza información restringida desde red externa. |
| 3 | RRHH consulta una licencia médica durante horario laboral | Rol RRHH Salud, necesidad de saber, caso, dispositivo administrado, propósito, vigencia | Permitir mínimo necesario | Cumple contexto y limita el alcance al expediente autorizado. |
| 4 | Administrador técnico intenta leer documento sensible | Rol técnico, clasificación, propósito, sesión y recurso | Denegar | Administrar infraestructura no equivale a leer datos de negocio. |
| 5 | Auditor solicita registros de acceso de una investigación aprobada | Rol auditor, aprobación, alcance, caso, horario y vista solicitada | Permitir vista acotada | La aprobación y el alcance justifican solo la evidencia necesaria. |
| 6 | Usuario intenta exportar información personal desde dispositivo no administrado | Rol, propósito, exportación, dispositivo, red, volumen y aprobación | Denegar | Exportar requiere contexto confiable y controles adicionales. |

## Decisión arquitectónica – RBAC, ABAC o combinación

### Respuesta

El diseño basado solo en el rol “RRHH” es demasiado amplio: una cuenta comprometida, un acceso externo o un cambio de puesto podrían exponer todo el dominio. RBAC resuelve permisos base por función; ABAC debe resolver sensibilidad, área, necesidad de saber, propósito, horario, red, dispositivo, sesión, aprobación y estado del documento. Leer, modificar y exportar son acciones distintas y deben tener permisos separados.

La combinación se refuerza con mínimo privilegio, sesiones cortas, MFA reforzado, límites de volumen, revisión trimestral, separación de funciones y revocación inmediata.

### Decisión del equipo

- [ ] Solo RBAC.
- [ ] Solo ABAC.
- [x] RBAC para permisos base y ABAC para restricciones contextuales.
- [ ] Otra combinación.

**Arquitectura seleccionada:** RBAC como autorización base y ABAC como filtro obligatorio antes del retrieval.

**Justificación:** Permite administrar permisos organizacionales sin perder precisión contextual ni trazabilidad. El LLM no participa en la decisión.

## Política de autorización previa a la recuperación

BankAssist HR permite una consulta solo cuando la identidad está autenticada, la cuenta activa, el rol autorizado para el dominio, la clasificación compatible con el nivel del usuario, los atributos contextuales permitidos, el propósito aprobado y el documento vigente/no revocado. La evaluación ocurre en el backend y el Policy Enforcement Point antes de seleccionar índices o recuperar chunks.

- **Información salarial:** solo Compensación/RRHH autorizado, necesidad de saber, dispositivo administrado, horario laboral, caso aprobado y sin exportación por defecto.
- **Licencias médicas:** solo RRHH Salud autorizado, expediente del caso permitido, propósito laboral legítimo, mínimo necesario y acceso auditado.
- **Políticas generales:** empleados, supervisores y RRHH pueden consultarlas desde una sesión válida; no contienen expedientes personales.
- **Fuera de horario:** información restringida se deniega o exige step-up MFA y aprobación; información interna de bajo riesgo puede permitirse con monitoreo reforzado.
- **Exportación:** nunca se hereda del permiso de lectura; exige aprobación, minimización, límite de volumen, watermark y registro auditable.

Se deniega por defecto ante identidad ausente, rol insuficiente, documento expirado/revocado, metadatos incompletos, contexto riesgoso, propósito no permitido o acceso a datos de otra persona sin justificación. Una denegación no consulta el RAG ni invoca el LLM.

## Preguntas reflexivas

### Reflexión 1

Pedir al LLM que no revele información no es autorización: el modelo puede seguir recibiendo el dato, ser manipulado por prompt injection, resumirlo indirectamente o fallar. La seguridad debe impedir que el dato no autorizado llegue al contexto y debe aplicar controles independientes del modelo.

### Reflexión 2

Un rol legítimo no autoriza automáticamente todo el dominio. Mínimo privilegio exige considerar propósito, necesidad de saber, clasificación, titularidad, caso, horario, red, dispositivo y riesgo. Leer, modificar y exportar requieren permisos distintos y revisiones periódicas.

## Sección 6 – Segmentación del RAG y separación por dominio

En un sistema RAG, la recuperación de contexto es un punto crítico de seguridad.

El sistema no debe buscar indistintamente en todos los documentos disponibles. La búsqueda debe limitarse según:

- La identidad del usuario.
- Su rol.
- Sus atributos.
- El dominio solicitado.
- La clasificación de la información.
- El propósito de la consulta.
- El estado y vigencia del documento.

### Principio arquitectónico

> La autorización debe aplicarse antes de recuperar el contexto, no después de que el LLM haya recibido la información.

La respuesta final del modelo no debe ser el mecanismo principal para ocultar datos. Si un documento no está autorizado, no debe llegar al contexto ni al prompt.

## Diseño de dominios e índices vectoriales

**Alternativa seleccionada: D – Índices separados más filtros obligatorios.**

BankAssist HR tendrá índices independientes para `rrhh_general`, `compensacion_beneficios`, `salud_ocupacional`, `relaciones_laborales`, `auditoria_cumplimiento` y `operaciones_internas`. Antes de consultar, el motor de políticas entrega al RAG una allow-list firmada con los índices autorizados y filtros de rol, atributos, clasificación, propósito, vigencia y revocación. El servicio RAG no puede ampliar esa lista.

La alternativa global sin filtros expone contexto al modelo; la global con filtros reduce riesgo pero conserva un radio de impacto mayor. La separación por dominio facilita aislamiento, mínimo privilegio, revocación, reindexación y auditoría. La similitud semántica nunca sustituye la autorización.

## Flujo de recuperación autorizada

1. Recibir consulta y validar esquema, tamaño, PII y prompt injection.
2. Identificar al usuario con SSO/MFA, cuenta activa y sesión vigente.
3. Obtener rol, área, nivel de autorización, red, dispositivo, horario y riesgo.
4. Determinar el dominio por intención, pero nunca permitir que la intención eleve permisos.
5. Evaluar RBAC + ABAC, propósito y clasificación máxima; denegar por defecto.
6. Entregar al RAG una allow-list firmada de índices autorizados.
7. Aplicar filtros previos de dominio, roles, atributos, vigencia, revocación y `top_k`.
8. Recuperar solo chunks con metadatos e integridad válidos.
9. Minimizar el contexto y eliminar campos innecesarios o datos de terceros.
10. Invocar el LLM mediante TLS/mTLS, secretos gestionados y proveedor sin entrenamiento.
11. Validar grounding, PII, alcance, instrucciones maliciosas y formato de respuesta.
12. Responder y registrar IDs, política, decisión, modelo y latencia, nunca prompt/contexto completo.

## Sección 7 – Aislamiento de entornos y separación de cargas

Una arquitectura segura de AI/LLM debe separar, como mínimo:

- Desarrollo.
- Pruebas.
- Validación o preproducción.
- Producción.

La separación no debe limitarse a una etiqueta o configuración lógica. Cada entorno debe tener controles propios y límites técnicos claros.

### Cada entorno debería definir

- Infraestructura independiente o claramente aislada.
- Credenciales distintas.
- Almacenamiento separado.
- Índices vectoriales separados.
- Redes o cuentas cloud diferenciadas.
- Políticas de acceso propias.
- Logs y monitoreo independientes.
- Variables y secretos gestionados de forma separada.
- Datos autorizados para ese entorno.

### Nunca debería permitirse

- Reutilizar índices vectoriales productivos en desarrollo.
- Utilizar datos personales reales en pruebas sin anonimización.
- Compartir secretos entre desarrollo, pruebas y producción.
- Permitir que una cuenta de pruebas invoque recursos productivos.
- Mezclar logs de producción con logs de desarrollo sin clasificación.
- Utilizar conversaciones reales para entrenamiento sin una justificación y control adecuados.

> El aislamiento de entornos reduce el impacto de errores, configuraciones incorrectas y credenciales comprometidas.

## Diseño de datos por entorno

Desarrollo usa únicamente datos sintéticos, anonimizados o documentos ficticios; se prohíben datos personales reales, índices productivos y secretos productivos. Pruebas usa casos controlados y datos anonimizados, con índices propios y cuentas de servicio de prueba. Preproducción usa una muestra limitada y autorizada, con formato cercano a producción, pero sin acceso de escritura a datos productivos. Producción usa datos reales autorizados, secretos gestionados, monitoreo activo, logging mínimo y controles de cambio.

El acceso temporal a datos reales en pruebas solo puede existir por excepción aprobada por privacidad, seguridad y propietario, con minimización, fecha de expiración, entorno aislado y evidencia de borrado.

## Matriz de separación de entornos

| Control | Desarrollo | Pruebas | Preproducción | Producción |
|---|---|---|---|---|
| Credenciales independientes | Completo | Completo | Completo | Completo con rotación |
| Índices vectoriales independientes | Completo | Completo | Completo | Completo por dominio |
| Datos anonimizados | Sintéticos | Obligatorio | Preferido/minimizado | No aplica a datos autorizados |
| Acceso a datos reales | Denegado | Denegado salvo excepción | Limitado y aprobado | Permitido según política |
| Acceso a proveedor LLM | Sandbox | Endpoint test con cuotas | Endpoint aprobado | Endpoint contratado y privado |
| Registro de actividad | Técnico | Técnico + pruebas | Auditable | SIEM/WORM |
| Aprobación de cambios | Revisión de equipo | QA + seguridad | CAB | CAB + propietario |
| Acceso de administradores | Sin contenido | Sin contenido | JIT | JIT, grabado y sin negocio |
| Retención de datos | Días | Ciclo de pruebas | Política controlada | Política legal/operativa |

## Separación de cargas sensibles

| Carga o tipo de dato | Entorno permitido | Ubicación separada | Acceso | Retención | Evidencia |
|---|---|---|---|---|---|
| Datasets de entrenamiento | Dev/test aprobados | Data lake no productivo | Gobierno de datos | Solo mientras exista propósito | Aprobación de privacidad |
| Documentos RAG | Preprod/prod según autorización | Object storage e índices por dominio | Propietario + RAG | Vigencia + política | Catálogo y hash |

## Caso de diseño – Reutilización de producción en pruebas

La propuesta se rechaza. Mezclar la base vectorial, credenciales y logs productivos expone PII, salarios, salud, prompts, respuestas y secretos; una consulta de prueba podría recuperar datos reales y una credencial comprometida permitiría movimiento lateral o extracción masiva. La alternativa es usar cuentas, redes, índices, almacenamiento, logs y secretos separados, con datos sintéticos/anonimizados y endpoints LLM de prueba con cuotas. Antes de cualquier excepción se exige aprobación de privacidad, seguridad y propietario, análisis de impacto, alcance mínimo, fecha de expiración, prueba de anonimización y evidencia de borrado. No se permite acceso directo permanente a producción.

## Decisión arquitectónica – Aislamiento mínimo aceptable

```text
Desarrollo:
  Cuenta/VPC/índice/log propios; datos sintéticos; secretos dev; sin rutas a producción.

Pruebas:
  Cuenta/VPC/índice propios; datos anonimizados; endpoint LLM de prueba; cuotas y DLP.

Preproducción:
  Cuenta y red propias; muestra autorizada; configuración equivalente; sin escritura en producción.

Producción:
  Cuenta/red/almacenamiento/índices por dominio; datos autorizados; Vault/KMS; SIEM y JIT.

Reglas de promoción:
  revisión de código, SAST/DAST, pruebas RBAC/ABAC/RAG, DLP, prompt injection,
  borrado, aislamiento, aprobación de seguridad, privacidad, plataforma y propietario.

Excepciones permitidas:
  solo copia minimizada/anonimizada, caso aprobado, acceso temporal, expiración
  automática y evidencia de eliminación; nunca credenciales o índices productivos.
```

## Investigación guiada 4 – Aislamiento y separación de cargas

1. Aislar ambientes significa separar cuentas, redes, identidades, almacenamiento, índices, secretos, datos y políticas para limitar el radio de impacto.
2. La separación lógica usa controles dentro de una plataforma compartida; la física usa cuentas, redes o infraestructura independientes y ofrece mayor resistencia a errores de configuración.
3. Credenciales compartidas permiten movimiento lateral y hacen imposible atribuir acciones o revocar un solo entorno.
4. Se usan anonimización irreversible, tokenización, enmascaramiento, generalización, datos sintéticos y minimización.
5. Pruebas usan endpoint LLM propio, allow-list, cuotas, TLS, no entrenamiento y logs separados.
6. IAM deny, VPC sin rutas, firewall, cuentas separadas, secretos propios y alertas SIEM evitan movimiento lateral.
7. Promoción exige revisión, SAST/DAST, pruebas de permisos/RAG/DLP, aprobación, rollback y evidencia de pipeline.

## Preguntas que podría hacer un auditor

Producción no comparte credenciales con pruebas: cada cuenta usa secretos y workload identity independientes. Un desarrollador no puede consultar índices productivos porque IAM, red privada y KMS lo impiden; los accesos administrativos son JIT, aprobados y registrados. Los secretos viven en un vault por entorno y no se copian mediante el pipeline. Copias de producción a pruebas están prohibidas salvo excepción aprobada, minimizada y anonimizada; DLP verifica el resultado. Backups y logs también están separados y tienen TTL propio. Las redes no tienen rutas laterales y el SIEM alerta intentos cross-environment. La promoción exige pruebas de seguridad, revisión de código, aprobación y evidencia de despliegue.

## Preguntas reflexivas

### Reflexión 1

Muchas filtraciones nacen en desarrollo o pruebas porque tienen controles más débiles, más personas con acceso, datos copiados sin anonimizar, secretos en código y redes con permisos amplios. La separación técnica y el uso de datos sintéticos reducen ese riesgo.

### Reflexión 2

El arquitecto debe buscar representatividad sin usar información real innecesaria. Se priorizan datos sintéticos, anonimización irreversible, minimización y muestras aprobadas. Si un caso excepcional exige datos reales, debe tener propósito, acceso temporal, controles equivalentes a producción, aprobación y borrado verificable.

## Sección 8 – Logging mínimo, monitoreo y trazabilidad segura

Los sistemas AI/LLM necesitan registrar actividad para:

- Diagnóstico técnico.
- Auditoría de uso.
- Detección de incidentes.
- Investigación forense.
- Cumplimiento regulatorio.
- Mejora controlada del servicio.

Sin embargo, los logs pueden contener información sensible si almacenan:

- Prompts completos.
- Respuestas completas.
- Contexto recuperado por RAG.
- Nombres, correos u otros datos personales.
- Información salarial o médica.
- Credenciales o tokens.
- Instrucciones internas del sistema.

### Principio de logging mínimo necesario

La arquitectura debe registrar suficiente información para reconstruir una operación, pero no más información de la necesaria.

> Trazabilidad no significa almacenar todo el contenido de la interacción.

## Qué registrar y qué evitar

| Elemento | ¿Debe registrarse? | Nivel de detalle | Justificación |
|---|---|---|---|
| Identificador de transacción | Sí | UUID | Correlación sin contenido |
| Identidad o identificador anonimizado | Sí | Hash estable | Investigación controlada |
| Rol del usuario | Sí | Valor de rol | Explica decisión |
| Dominio consultado | Sí | Nombre de dominio | Trazabilidad |
| Documento utilizado | Sí | ID, versión y chunk | Reconstruye fuente sin texto |
| Contenido completo del documento | No | Ninguno | Evita fuga |
| Prompt completo | No | Ninguno | Puede contener PII/injection |
| Respuesta completa | No | Ninguno | Puede contener datos sensibles |
| Código de estado | Sí | HTTP/estado lógico | Operación |
| Tiempo de respuesta | Sí | Milisegundos | Rendimiento |
| Modelo utilizado | Sí | ID y versión | Reproducibilidad |
| Versión de la política de acceso | Sí | ID/version | Auditoría |
| Error técnico | Sí | Código sanitizado | Diagnóstico sin secretos |
| Token o secreto | No | Nunca | Riesgo crítico |

## Separación de logs técnicos y registros auditables

BankAssist HR debe diferenciar entre dos categorías de registros.

### Logs técnicos

Se utilizan para operación y diagnóstico:

- Rendimiento.
- Errores.
- Disponibilidad.
- Latencia.
- Uso de CPU, memoria o red.
- Estado de componentes.
- Fallas de conexión.

### Registros auditables

Se utilizan para reconstruir decisiones y accesos:

- Quién realizó la consulta.
- Qué dominio solicitó.
- Qué política se aplicó.
- Si la solicitud fue permitida o denegada.
- Qué documentos o referencias se utilizaron.
- Qué versión del modelo y de la política estaba activa.
- Qué acción crítica se ejecutó.

### Ambos tipos de registros deben tener

- Cifrado.
- Control de acceso diferenciado.
- Retención limitada.
- Integridad protegida.
- Sin secretos ni contenido innecesario.
- Sin acceso general para todos los administradores.

## Diseño de un evento auditable

```json
{
  "event_id": "evt-uuid",
  "timestamp": "2026-08-17T12:00:00Z",
  "environment": "production",
  "user_id_hash": "sha256:...",
  "user_role": "empleado",
  "request_id": "req-uuid",
  "requested_domain": "rrhh_general",
  "authorization_decision": "allow",
  "policy_version": "auth-2026.08",
  "retrieved_document_ids": ["hr-policy-001:v3:chunk-04"],
  "model_id": "bankassist-llm-prod",
  "response_status": "success",
  "sensitive_content_stored": false,
  "retention_class": "audit-12m"
}
```

Los campos necesarios permiten correlacionar la solicitud, conocer quién/qué rol/qué dominio/política/resultado intervino y reconstruir las referencias usadas. `user_id` debe anonimizarse; prompts, respuestas, contexto, PII, tokens, secretos y nombres completos se omiten. Los IDs de documentos, versión, timestamps y decisión permiten investigar sin exponer contenido. El evento se cifra, firma o encadena en un almacén con acceso separado.

## Caso de diseño – Extracción masiva

Se correlacionan volumen (1.500 consultas), dominios sensibles, red nueva, exportaciones y respuestas bloqueadas con identidad, IP, dispositivo, sesión y cambios de permisos. API/RAG y SIEM generan la alerta crítica. Se aplican automáticamente rate limit, revocación de sesión/token, bloqueo de exportaciones, step-up MFA y aislamiento temporal; seguridad decide suspensión definitiva y alcance forense.

Se conservan request IDs, timestamps, usuario hash, dominios, IDs/versiones de documentos, decisiones, políticas, IP normalizada, alertas y acciones, nunca prompts, respuestas ni contexto. La investigación usa vistas enmascaradas y acceso JIT. El patrón indica credencial comprometida, automatización abusiva o insider threat; se rota el secreto, se revisan accesos y se preserva evidencia con integridad.

## Sección 9 – Retención, borrado y ciclo de vida de los datos

Una aplicación AI/LLM puede generar y almacenar información en múltiples ubicaciones:

- Documentos originales.
- Fragmentos utilizados por RAG.
- Embeddings.
- Índices vectoriales.
- Historiales de conversación.
- Prompts y respuestas.
- Logs técnicos.
- Registros de auditoría.
- Backups.
- Caches temporales.
- Datasets derivados.
- Métricas de uso.

La organización no debe conservar estos datos indefinidamente sin un propósito definido.

### Cada categoría de datos debe tener

- Un propósito de uso.
- Un propietario.
- Una clasificación.
- Un período máximo de conservación.
- Una base legal, regulatoria u operativa.
- Una acción al vencer el período.
- Un proceso de revisión o eliminación.
- Evidencia verificable.

## Matriz de retención por tipo de dato

| Tipo de dato | Propósito | Propietario | Sensibilidad | Retención propuesta | Acción al vencer | Evidencia |
|---|---|---|---|---|---|---|
| Documentos originales | Fuente autorizada | Propietario de dominio | Según fuente | Vigencia + 24 meses | Archivar/eliminar | Catálogo y purga |
| Fragmentos RAG | Retrieval | Propietario/RAG | Igual a fuente | Igual a fuente | Borrar y reindexar | IDs y reporte |
| Embeddings | Búsqueda semántica | Propietario/RAG | Igual a fuente | Igual a fuente | Invalidar/eliminar | Conteo antes/después |
| Historiales de conversación | Servicio al usuario | Privacidad | Personal | 30 días | Purga automática | TTL |
| Prompts | Inferencia | Privacidad | Personal | No persistir; 0 días | No almacenar | Prueba de configuración |
| Respuestas generadas | Presentación | Privacidad | Personal | No persistir; 0 días | No almacenar | DLP |
| Logs técnicos | Diagnóstico | Operaciones | Interna | 30 días | Purga | Job y reporte |
| Registros auditables | Cumplimiento | Seguridad | Confidencial | 12 meses/legal | Purga controlada | WORM y certificado |
| Backups | Recuperación | Plataforma | Heredada | Máximo 35 días | Expirar/destruir clave | Inventario |
| Caches temporales | Rendimiento | Plataforma | Heredada | Minutos/horas | Expirar inmediatamente | TTL y prueba |

## Borrado seguro y eliminación verificable

El borrado de un dato no debe limitarse al repositorio principal.

Cuando una información deja de conservarse o debe eliminarse, la arquitectura debe revisar:

- Almacenamiento primario.
- Bases de datos de aplicación.
- Índices vectoriales.
- Embeddings.
- Caches.
- Historiales de conversación.
- Logs que contengan referencias o contenido.
- Backups.
- Datasets derivados.
- Réplicas.
- Sistemas de terceros, si corresponde.

### Posibles mecanismos

- Eliminación lógica controlada.
- Eliminación física.
- Destrucción o invalidación criptográfica.
- Revocación de claves.
- Eliminación de referencias.
- Expiración automática.
- Reindexación.
- Destrucción de caches.
- Procesos de purga en backups.

### Advertencia

Eliminar un documento original no garantiza automáticamente que haya desaparecido de:

- El índice vectorial.
- Los embeddings.
- Las respuestas almacenadas.
- Los logs.
- Las copias de respaldo.

## Flujo de eliminación de datos

1. Privacidad recibe la solicitud autenticada por SSO y verificación adicional.
2. Se determina el alcance: usuario, conversaciones, documentos y derivados.
3. El inventario localiza base principal, historial, caches, RAG, embeddings, logs, backups y proveedores.
4. Legal/privacidad documenta excepciones de conservación y su plazo.
5. Se elimina el contenido en la base y se bloquean nuevas ingestas.
6. Se borran chunks, embeddings, índices, caches, historiales y copias derivadas; se notifica al proveedor.
7. Se verifica con búsquedas por ID, reindexación y prueba adversarial que ya no sea recuperable.
8. Se conserva solo un comprobante hash con alcance, componentes, timestamps, operador y resultado.
9. Privacidad comunica al solicitante lo eliminado, las excepciones y la fecha de finalización.

Responsables: privacidad coordina; propietarios validan alcance; plataforma ejecuta repositorios/backups; RAG reindexa; seguridad verifica evidencia; legal aprueba excepciones.

## Gestión de solicitudes de privacidad

En entornos regulados, la arquitectura debe permitir gestionar solicitudes de:

- Acceso a datos personales.
- Rectificación.
- Eliminación.
- Restricción del tratamiento.
- Revisión de información almacenada.
- Información sobre el propósito de uso.

- Sistemas involucrados.
- Responsable de la solicitud.

- Respuesta al solicitante.

## Caso de diseño – Documento eliminado, embedding activo

1. Deben eliminarse el original, chunks, embedding, índice, cache, historial y cualquier copia derivada; backups se purgan según ciclo o destrucción de clave.
2. Solo puede conservarse evidencia técnica mínima: hash de solicitud, timestamps, componentes, estado y operador. El ID debe anonimizarse.
3. El contenido eliminado no se conserva; el registro demuestra que la operación ocurrió sin reproducir el dato.
4. Se revoca el documento, se elimina el vector y se reindexa el dominio; el filtro excluye el ID mientras termina el proceso.
5. Se ejecutan búsquedas por ID, consultas semánticas adversariales, verificación de cache y conteo de vectores.
6. Se comunica alcance, sistemas tratados, resultado, excepciones legales y fecha de cierre, sin mostrar el contenido borrado.

## Investigación guiada 6 – Ciclo de vida y privacidad

1. Minimización, limitación de propósito, necesidad operativa y privacidad justifican limitar conservación.
2. Retención es plazo definido; archivado conserva con acceso reducido; eliminación hace que el dato deje de estar disponible.
3. Sistemas distribuidos tienen réplicas, caches, índices, backups y proveedores que deben coordinarse.
4. Backups pueden prolongar disponibilidad; deben tener TTL, cifrado y purga o destrucción de clave.
5. Embeddings y caches son derivados recuperables y deben borrarse junto con la fuente.
6. Inventario, borrado por ID, reindexación, búsquedas negativas, conteos y certificados verifican efectividad.
7. Puede conservarse hash de solicitud, timestamp, alcance, componentes, operador y resultado, sin contenido.
8. Conversaciones y logs pueden contener datos personales y requieren propósito, base legal, acceso, retención y borrado.

## Flujo seguro de una consulta

Para “¿Cuántos días de vacaciones me corresponden según mi antigüedad?” se valida identidad SSO/MFA, cuenta activa, rol, área, propósito, red, dispositivo y riesgo. Se selecciona `rrhh_general` y solo índices/documentos de políticas vigentes permitidos al empleado. Se excluyen expedientes, salarios, licencias y datos de otros empleados. El contexto mínimo se envía por TLS/mTLS al modelo aprobado; la respuesta se valida contra las referencias y con DLP. Se registra request ID, usuario hash, dominio, documentos/versiones, política, modelo, estado y latencia, nunca prompt/respuesta completa. Una anomalía activa rate limit/SIEM y el historial se conserva 30 días; el registro auditable, 12 meses o legal.

## Flujo seguro de una consulta denegada

1. Autenticarse no basta: identifica al usuario, pero no concede el permiso.
2. El usuario tiene rol empleado común.
3. Intervienen `compensacion_beneficios` y `salud_ocupacional`.
4. Salario y licencia médica son datos restringidos/sensibles de terceros.
5. La política deniega por rol, propósito y titularidad del dato.
6. No se ejecuta búsqueda en el RAG.
7. No se envía nada al LLM.
8. Se responde: “No tienes permisos para consultar esa información”.
9. Se registra usuario hash, dominio, motivo, política, timestamp y `model_invoked=false`, sin contenido.
10. Repetición puede generar alerta por intentos denegados, rate limit, step-up y revisión SOC.

# Solución aplicada – BankAssist HR

> Esta sección consolida las respuestas arquitectónicas para todas las preguntas y tablas del Lab 2. La decisión base es **deny by default**: el LLM nunca autoriza, nunca recibe documentos fuera del alcance del usuario y nunca se utiliza como filtro de seguridad.

## 1. Flujo y decisiones iniciales

El flujo aprobado es: **SSO/MFA -> API Gateway -> identidad y atributos -> Policy Enforcement Point -> selección de dominio -> filtros obligatorios -> recuperación RAG -> minimización y validación del contexto -> LLM privado o con garantía contractual de no entrenamiento -> validación de salida -> respuesta -> logging mínimo/SIEM**.

La autenticación identifica al empleado; RBAC concede permisos base; ABAC evalúa área, clasificación, propósito, red, dispositivo, horario y riesgo; el RAG solo consulta índices permitidos. Las denegaciones ocurren antes del retrieval y generan un evento con `request_id`, usuario anonimizado, dominio, política, motivo y `model_invoked=false`.

| Riesgo | Decisión | Evidencia |
|---|---|---|
| Fuga vía RAG | Índices separados por dominio y filtros previos por rol, clasificación, propósito y vigencia | Decisión de política, IDs de chunks y pruebas negativas |
| Acceso indebido | RBAC + ABAC con denegación por defecto | Matriz versionada y logs de autorización |
| Datos reales en pruebas | Cuentas, redes, secretos, almacenamiento e índices separados; datos sintéticos o anonimizados | Inventario de entornos y prueba de aislamiento |
| PII en logs | No guardar prompts, respuestas ni contenido; usar hashes y referencias | Esquema de evento y escaneo DLP |
| Retención indefinida | TTL por categoría, purga de derivados y verificación | Jobs de borrado, reportes y certificados |
| Credenciales comprometidas | Workload identity, vault, rotación, MFA y JIT | Evidencia IAM, rotaciones y alertas |

## 2. Clasificación y catálogo de datos

| Tipo | Dominio | Clasificación | Uso y acceso | Retención |
|---|---|---|---|---|
| Política general de vacaciones | `rrhh_general` | Interna | Consulta de empleados, supervisores y RRHH | 24 meses tras sustitución |
| Procedimiento interno de RRHH | `rrhh_general` | Confidencial | RRHH; supervisores solo si el procedimiento lo permite | Vigencia + 24 meses |
| Licencia médica individual | `salud_ocupacional` | Restringida/sensible | RRHH Salud autorizado y caso justificado | Según obligación legal; luego purga |
| Información salarial | `compensacion_beneficios` | Restringida | Compensación y RRHH autorizado; nunca empleado común | Según política laboral y legal |
| Conversación | Aplicación | Personal/confidencial | Solo el titular y servicio controlado | 30 días, salvo retención aprobada |
| Documento y fragmento RAG | Dominio de origen | Igual o superior al original | Solo sujetos autorizados | Hasta expiración del documento |
| Embedding | Dominio de origen | Igual a la fuente | Solo servicio RAG y administradores sin acceso al contenido | Se elimina con la fuente |
| Log técnico | Observabilidad | Interna, potencialmente personal | Soporte acotado | 30 días |
| Registro auditable | Auditoría | Confidencial | Auditoría, seguridad y privacidad con caso aprobado | 12 meses o exigencia legal |

**Metadatos obligatorios:** `document_id`, `chunk_id`, `domain`, `data_owner`, `classification`, `allowed_roles`, `allowed_attributes`, `purpose`, `source_system`, `created_at`, `updated_at`, `expires_at`, `version`, `embedding_version`, `retention_class`, `revocation_status` y `integrity_hash`. Solo el propietario del dato, mediante flujo aprobado, puede cambiar clasificación o permisos; cada cambio requiere revisión por seguridad y queda versionado.

Un embedding, cache o log se trata como el dato de origen o con una clasificación más restrictiva. Que un embedding no sea legible no elimina el riesgo de inferencia, recuperación, reconstrucción o correlación.

## 3. RBAC, ABAC y política de autorización

| Recurso | Empleado | Supervisor | RRHH | Auditor | Admin técnico |
|---|---|---|---|---|---|
| Política general | Permitido | Permitido | Permitido | Solo metadatos | Denegado al contenido |
| Procedimiento RRHH | Denegado o condicionado | Condicionado | Permitido | Solo metadatos | Denegado |
| Salarios | Denegado | Denegado salvo caso aprobado | Permitido con necesidad de saber | Solo evidencia autorizada | Denegado |
| Licencias médicas | Denegado | Denegado | Permitido con caso y atributo | Solo evidencia | Denegado |
| Logs técnicos | Denegado | Denegado | Denegado | Según caso | Solo operación, sin PII |
| Auditoría | Denegado | Denegado | Según función | Permitido con aprobación | Solo metadatos técnicos |
| Modificar documentos | Denegado | Denegado | Permitido al propietario | Denegado | No contenido |
| Modificar clasificación | Denegado | Denegado | Proponer | Aprobar/revisar | Denegado |
| Exportar | Denegado por defecto | Requiere aprobación | Condicionado y auditado | Caso aprobado | Denegado |

**Reglas ABAC:** política general: cuenta activa, red corporativa o MFA reforzado, propósito `consulta_politicas`; salario: rol autorizado, relación laboral permitida, dispositivo administrado, horario laboral, caso aprobado y no exportación; licencia médica: RRHH Salud, caso justificado, propósito laboral legítimo y acceso individual mínimo; auditoría: caso aprobado, alcance limitado y vista enmascarada; administrador técnico: solo salud del servicio y metadatos técnicos.

Fuera de horario se deniega información restringida y se permite únicamente información interna de bajo riesgo con MFA y monitoreo reforzado. Leer no implica exportar: la exportación exige propósito, aprobación, minimización, watermark, límite de volumen y registro.

La arquitectura seleccionada es **RBAC para permisos base + ABAC para restricciones contextuales**. Solo roles sería demasiado amplio y una cuenta comprometida tendría un radio de impacto excesivo. Se reduce el riesgo con sesiones cortas, step-up MFA, límites de volumen, separación de funciones, revisión periódica y revocación inmediata.

## 4. Segmentación y contrato RAG

La alternativa elegida es **índices separados por dominio y filtros obligatorios previos**: `rrhh_general`, `compensacion_beneficios`, `salud_ocupacional`, `relaciones_laborales`, `auditoria_cumplimiento` y `operaciones_internas`. El selector recibe una allow-list firmada por el motor de políticas; el servicio RAG no puede ampliar esa lista.

**Contrato:**

- Entrada: identidad verificada, rol, atributos, propósito, dominio solicitado, clasificación máxima, `request_id` y versión de política.
- Validaciones: cuenta activa, sesión válida, dominio permitido, metadatos completos, documento vigente/no revocado, filtros ABAC y límites de volumen.
- Recuperación: aplicar filtros de autorización antes de la similitud; consultar solo índices de la allow-list; limitar `top_k`; excluir chunks sin integridad o metadatos.
- Salida permitida: chunks mínimos autorizados, `document_id`, `chunk_id`, versión, dominio y score técnico.
- Salida prohibida: documentos fuera de dominio, expirados, revocados, contenido no autorizado, secretos, prompt del sistema y datos de otros empleados.
- Evidencia: IDs y versiones recuperados, hash del filtro, política aplicada, decisión, latencia y `model_invoked`.

Una consulta sobre el salario del gerente se dirige a `compensacion_beneficios`, pero un empleado común no está autorizado: se rechaza antes del RAG, no se llama al LLM y se responde de forma neutra: “No tienes permisos para consultar esa información”. Se registra el intento sin repetir el contenido.

## 5. Entornos y cargas

| Entorno | Cuenta/red/índice | Datos | Credenciales y LLM |
|---|---|---|---|
| Desarrollo | Cuenta, VPC y vector DB propias | Sintéticos o anonimizados | Secretos de desarrollo; proveedor sandbox |
| Pruebas | Aislado de desarrollo y producción | Dataset controlado anonimizado | Cuenta de pruebas; cuotas y allow-list |
| Preproducción | Espejo aislado, sin ruta de escritura a producción | Muestra aprobada y minimizada | Credenciales propias; pruebas de seguridad |
| Producción | Cuenta/red, índices y almacenamiento dedicados | Datos reales autorizados | Vault/KMS, identidad de workload, rotación y LLM aprobado |

Nunca se comparte índice, secreto, backup, log o credencial entre entornos. Una promoción exige revisión de código, SAST/DAST, pruebas de autorización permitida/denegada, prueba de prompt injection, escaneo de secretos y aprobación de seguridad, privacidad y propietario del dato. La única excepción posible es una copia temporal, minimizada y aprobada, con expiración automática y trazabilidad; no se permite reutilizar producción directamente.

Las cargas se separan por propósito: documentos y embeddings en repositorios RAG por dominio; conversaciones en almacenamiento personal con TTL; logs técnicos y auditoría en almacenes separados; caches cifradas y de vida corta; datasets de entrenamiento prohibidos por defecto y solo autorizables tras revisión de privacidad.

## 6. Logging, monitoreo y SIEM

**Registrar:** `event_id`, timestamp, entorno, `request_id`, hash de usuario, rol, dominio, decisión, motivo codificado, versión de política, IDs/versiones de documentos, modelo, estado, latencia, error normalizado, volumen y retención. **Excluir:** prompt y respuesta completos, contexto, PII, salarios, salud, tokens, claves y secretos. Los logs técnicos duran 30 días; los auditables 12 meses o el plazo legal, con cifrado, control separado, integridad encadenada y acceso just-in-time.

```json
{
  "event_id": "evt-uuid",
  "timestamp": "2026-08-17T12:00:00Z",
  "environment": "production",
  "user_id_hash": "sha256:...",
  "user_role": "empleado",
  "request_id": "req-uuid",
  "requested_domain": "rrhh_general",
  "authorization_decision": "allow",
  "policy_version": "auth-2026.08",
  "retrieved_document_ids": ["hr-policy-001:v3:chunk-04"],
  "model_id": "bankassist-llm-prod",
  "response_status": "success",
  "sensitive_content_stored": false,
  "retention_class": "audit-12m"
}
```

Alertas: más de 100 consultas en 10 minutos; 5 denegaciones sensibles en 5 minutos; exportación o acceso a 3 dominios restringidos en una hora; acceso simultáneo desde ubicaciones incompatibles; lectura directa del índice o cambio de metadatos fuera del flujo aprobado. Las acciones iniciales son rate limit, step-up MFA, bloqueo temporal de exportación, revocación de sesión y ticket SIEM; la suspensión permanente y la investigación requieren intervención humana.

En el caso de 1.500 consultas, salud, compensación, red nueva, exportaciones y bloqueos, se correlacionan API, IAM, RAG, DLP, red y SIEM. Se congela la sesión, se revoca el token, se preservan metadatos y hashes, se bloquean exportaciones y se abre incidente. La investigación usa IDs y referencias, nunca el contenido sensible.

## 7. Retención y borrado

| Dato | Retención | Acción al vencer |
|---|---|---|
| Documento original | Vigencia + 24 meses | Archivar si obligación; si no, eliminar |
| Fragmento/embedding/índice | Igual al documento | Borrar, reindexar y verificar ausencia |
| Conversación/prompt/respuesta | 30 días | Purga automática |
| Cache | Minutos u horas | Expiración y destrucción |
| Log técnico | 30 días | Purga |
| Registro auditable | 12 meses o legal | Purga con excepción documentada |
| Backup | Ciclo máximo 35 días | Purga criptográfica o expiración |

Una solicitud de privacidad se autentica por SSO y verificación adicional; el servicio de privacidad crea un inventario de datos por identificador; el propietario y legal determinan excepciones; se eliminan base principal, historial, caches, fragmentos, embeddings, índices y copias; se notifica al proveedor; backups se purgan al siguiente ciclo o mediante destrucción de clave cuando sea apropiado. Se prueba la no recuperación con búsquedas por ID, reindexación y consultas adversariales. Se conserva solo un comprobante mínimo: solicitud hash, alcance, timestamps, componentes, estado, operador y resultado, sin el dato eliminado.

En el caso de documento eliminado con embedding, cache, historial, log y backup activos, se elimina contenido y derivados, se invalida el embedding inmediatamente, se purga cache e historial, se minimiza el log a un identificador hash y se espera o ejecuta la purga del backup. El solicitante recibe alcance, resultado, excepciones y fecha de finalización.

| Etapa | Control | Evidencia |
