## Sección 1 – Contexto arquitectónico de BankAssist HR

BankAssist HR es un asistente interno para empleados de un banco latinoamericano.

El sistema permite consultar información relacionada con:

- Políticas de vacaciones y licencias.
- Beneficios y prestaciones sociales.
- Procedimientos internos de Recursos Humanos.
- Manuales operativos básicos.
- Información organizacional autorizada.

La solución utiliza una arquitectura basada en recuperación aumentada por generación, o **RAG**:

1. El empleado realiza una pregunta desde la intranet.
2. El backend autentica al usuario.
3. El sistema determina su rol y atributos de acceso.
4. El módulo RAG busca información únicamente en dominios autorizados.
5. El sistema filtra los documentos según su clasificación.
6. El contexto permitido se envía al LLM.
7. El modelo genera una respuesta.
8. La respuesta se valida antes de mostrarse.
9. La operación se registra mediante logging mínimo y seguro.

### Restricción principal

El sistema no debe enviar todos los documentos disponibles al LLM para intentar filtrar la información después.

La autorización debe aplicarse antes de:

- Recuperar documentos.
- Construir el contexto.
- Construir el prompt.
- Invocar el modelo.

> El modelo debe operar sobre un subconjunto de datos previamente autorizado por la arquitectura.

## Sección 5 – Diseño de control de acceso

En una aplicación AI/LLM, el control de acceso debe determinar qué información puede llegar al sistema de inferencia.

No es suficiente con controlar únicamente el acceso al portal. También deben protegerse:

- Los documentos originales.
- Los fragmentos utilizados por el RAG.
- Los índices vectoriales.
- Los embeddings.
- El contexto recuperado.
- Las conversaciones.
- Los logs y registros de auditoría.
- Las herramientas internas, si el sistema incorpora agentes.

### Principio de mínimo privilegio

Cada usuario, servicio o componente debe tener únicamente los permisos necesarios para cumplir su función.

En BankAssist HR, el sistema debe:

1. Autenticar al usuario.

## RBAC – Control basado en roles

RBAC (*Role-Based Access Control*) asigna permisos según el rol organizacional del usuario.

### Roles iniciales de BankAssist HR

- **Empleado:** consulta políticas generales autorizadas.
- **Supervisor:** consulta políticas generales y ciertos procedimientos de supervisión.
- **RRHH:** consulta y administra información de Recursos Humanos según sus funciones.
- **Auditor:** consulta evidencias y registros autorizados, pero no necesariamente el contenido completo de todos los documentos.
- **Administrador técnico:** administra componentes técnicos, pero no debería obtener automáticamente acceso al contenido de negocio.
- **Servicio RAG:** recupera documentos según las políticas evaluadas; no decide por sí mismo qué usuario tiene acceso.

### Advertencia arquitectónica

Un rol técnico no debe equivaler automáticamente a un permiso para consultar información sensible.

Por ejemplo, un administrador de infraestructura podría administrar una base de datos, pero no debería tener acceso ilimitado al contenido de licencias médicas o información salarial.

## Sección 6 – Segmentación del RAG y separación por dominio

En un sistema RAG, la recuperación de contexto es un punto crítico de seguridad.

El sistema no debe buscar indistintamente en todos los documentos disponibles. La búsqueda debe limitarse según:

- La identidad del usuario.
- Su rol.
- Sus atributos.
- El dominio solicitado.
- La clasificación de la información.
- El propósito de la consulta.
- El estado y vigencia del documento.

### Principio arquitectónico

> La autorización debe aplicarse antes de recuperar el contexto, no después de que el LLM haya recibido la información.

La respuesta final del modelo no debe ser el mecanismo principal para ocultar datos. Si un documento no está autorizado, no debe llegar al contexto ni al prompt.

## Sección 7 – Aislamiento de entornos y separación de cargas

Una arquitectura segura de AI/LLM debe separar, como mínimo:

- Desarrollo.
- Pruebas.
- Validación o preproducción.
- Producción.

La separación no debe limitarse a una etiqueta o configuración lógica. Cada entorno debe tener controles propios y límites técnicos claros.

### Cada entorno debería definir

- Infraestructura independiente o claramente aislada.
- Credenciales distintas.
- Almacenamiento separado.
- Índices vectoriales separados.
- Redes o cuentas cloud diferenciadas.
- Políticas de acceso propias.
- Logs y monitoreo independientes.
- Variables y secretos gestionados de forma separada.
- Datos autorizados para ese entorno.

### Nunca debería permitirse

- Reutilizar índices vectoriales productivos en desarrollo.
- Utilizar datos personales reales en pruebas sin anonimización.
- Compartir secretos entre desarrollo, pruebas y producción.
- Permitir que una cuenta de pruebas invoque recursos productivos.
- Mezclar logs de producción con logs de desarrollo sin clasificación.
- Utilizar conversaciones reales para entrenamiento sin una justificación y control adecuados.

> El aislamiento de entornos reduce el impacto de errores, configuraciones incorrectas y credenciales comprometidas.

- Clasificación de datos.
- Control de acceso.
- Segmentación por dominio.
- Filtrado previo del contexto.
- Protección de datos.
- Gestión de secretos.

- Monitoreo.
- Retención y borrado.

6. Recuperación filtrada de contexto.

8. Invocación cifrada del modelo.
9. Validación de la respuesta.
10. Entrega al usuario.
11. Logging mínimo.
12. Monitoreo.
13. Aplicación de retención y borrado.

Se trabajó sobre:

- Clasificación y gobierno de datos.
- RBAC y ABAC.
- Autorización antes de la recuperación.
- Segmentación de índices vectoriales.
- Metadatos de clasificación.
- Aislamiento de entornos.
- Separación de cargas sensibles.
- Logging mínimo.
- Monitoreo y detección de anomalías.
- Retención y borrado.
- Solicitudes de privacidad.
- Trazabilidad y auditoría.
- Integración end to end de controles.

La principal conclusión es:

> Una aplicación AI/LLM no es segura solo porque el modelo genere respuestas correctas. Es segura cuando la arquitectura controla qué datos utiliza, quién puede acceder, qué información llega al modelo, cómo se registra la operación, cuánto tiempo se conservan los datos y cómo se demuestra todo ante una auditoría.
